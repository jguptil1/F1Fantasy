```python
import pandas as pd
import requests
import lxml
from bs4 import BeautifulSoup
from io import StringIO
```

```python
url = "https://f1fantasytools.com/statistics"
```

```python
def test_request():
    headers = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/151.0.0.0 Safari/537.36"
        )
    }

    response = requests.get(url, headers=headers, timeout=30)

    print("Status code:", response.status_code)
    print("HTML length:", len(response.text))
    print()
    print(response.text[:1000])

    response.raise_for_status()

    return response.text
```

```python
def test_html_tables(html):
    try:
        tables = pd.read_html(StringIO(html))

        print(f"\nFound {len(tables)} HTML tables")

        for i, table in enumerate(tables):
            print(f"\n--- TABLE {i} ---")
            print(table.head())
            print(table.columns.tolist())

    except ValueError:
        print("\nNo HTML tables found.")
```

```python
def inspect_html(html):
    soup = BeautifulSoup(html, "html.parser")

    print("\nPage title:")
    print(soup.title)

    print("\nNumber of <table> elements:")
    print(len(soup.find_all("table")))

    print("\nScripts:")
    for script in soup.find_all("script"):
        src = script.get("src")

        if src:
            print(src)
```

```python
def inspect_embedded_data(html):
    keywords = [
        "verstappen",
        "norris",
        "leclerc",
        "fantasy",
        "price",
        "points",
        "driver",
        "constructor",
    ]

    html_lower = html.lower()

    print("\n--- KEYWORD CHECK ---")

    for keyword in keywords:
        count = html_lower.count(keyword.lower())
        print(f"{keyword}: {count}")
```

```python
html = test_request()

test_html_tables(html)
inspect_html(html)
inspect_embedded_data(html)
```
